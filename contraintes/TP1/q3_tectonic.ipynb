## Les règles

Le **Tectonic** (aussi appelé *Suguru*) se joue sur une grille découpée en **régions** de 1 à 5 cases, délimitées par des traits épais.
Certaines cases contiennent déjà un chiffre.
Il faut remplir toutes les cases en respectant deux règles :

1. une région de *t* cases contient **chacun des chiffres de 1 à *t*** exactement une fois
(une région de 3 cases contient 1, 2 et 3) ;
2. deux cases **voisines**, horizontalement, verticalement **ou en diagonale**, ne contiennent jamais le même chiffre, même si elles sont de régions différentes.
