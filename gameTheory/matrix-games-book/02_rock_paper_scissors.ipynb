## Exercise

Let's study the classical "Rock-Paper-Scissors" game: 2 players choose a sign simultaneously "Rock" , "Paper", or "Scissors".

- "Rock" wins against "Scissors", 
- "Scissors" wins against "Paper", and 
- "Paper" wins against "Rock". 

- If both players choose the same option, they receive 0 point; 
- otherwise, the winner gets +1 and the loser -1.

here ios the matrix game

| **A** \ B | rock | paper | scissors |
|---|---|---|---|
| **rock** | (**0**, 0) | (**−1**, 1) | (**1**, −1) |
| **paper** | (**1**, −1) | (**0**, 0) | (**−1**, 1) |
| **scissors** | (**−1**, 1) | (**1**, −1) | (**0**, 0) |

```python
#in Nashpy
acts = ['rock', 'paper', 'scissors']
A = np.array([[0, -1, 1], [1, 0, -1], [-1, 1, 0]])
B = -A
game = nash.Game(A, B)   # zero-sum: B = -A
```

**Q1.** Find the Nash equilibria in pure strategies.

> **Solution.**   
> **None**: best responses cycle (R → P → S → R …).  
> OR another explanation:  
> for A : best cells in each column = (1,0), (2,1), (0,2)   
> for B : best cells in each row = (0,1), (1,2), (2,0)  
> $bestCellsA \cap bestCellsB = \emptyset $  

**Q2.** Find the mixed-strategy equilibrium.

> **Solution.**   
>  With $p_r, p_p, p_s$ for A, B must be indifferent between R, P and S:  
>  $-p_p + p_s =  p_r  - p_s = p_r + p_p$  
>   with $p_s = 1 - p_p - p_r$ \Rightarrow $1 -2p_p - p_r = -1 + 2p_r + p_p  =  p_r + p_p$  
> \Rightarrow $p_r = 1/3$,  $p_p = 1/3$, $p_s = 1/3$,  
>  Same for B;   
> and the expected values for the both at the equilibrium is 0
> 
> *In a zero-sum game, making your own payoff independent of the other's choice is the same as making the other indifferent.*

**Check with nashpy**
