| A \ B | rock | paper | scissors |
|---|---|---|---|
| **rock** | (0, 0) | (−1, 1) | (1, −1) |
| **paper** | (1, −1) | (0, 0) | (−1, 1) |
| **scissors** | (−1, 1) | (1, −1) | (0, 0) |

```python
acts = ['rock', 'paper', 'scissors']
game = nash.Game(np.array([[0, -1, 1], [1, 0, -1], [-1, 1, 0]]))   # zero-sum: B = -A
```

**Q1.** Find the Nash equilibria in pure strategies.

**Solution.** **None**: best responses cycle (R → P → S → R …).

**Q2.** Find the mixed-strategy equilibrium.

**Solution.** With $p_r, p_p, p_s$ for A, B must be indifferent between R, P and S:
$2p_p + p_r - 1 = 1 - 2p_r - p_p = p_r - p_p \Rightarrow p_r = p_p = p_s = 1/3$. Same for B; value 0.

*In a zero-sum game, making your own payoff independent of the other's choice is the same as making the other indifferent.*

**Check with nashpy**

```python
print_equilibria(game, acts, acts)
```

**Q3.** What do you conclude?

**Solution.** The only equilibrium is to be unpredictable: any bias (e.g. more Rock) can be exploited (by Paper).
