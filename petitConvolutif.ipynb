# Réseau convolutif

Exemple simple de reseau convolutif ou reseau de neurones convolutionnel avec tensorflow et keras
- pour reconnaître des chiffres écrits a la main
- avec le dataset MNIST qui contient 60000 images d'entrainement et 10000 images de test

```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import matplotlib.pyplot as plt
import numpy as np
```

```python
# chargement des images et labels du dataset MNIST
(train_images, train_labels), (test_images, test_labels) = datasets.mnist.load_data()
```

```python
def drawImage(i):
    plt.figure()
    plt.imshow(train_images[i], cmap=plt.cm.binary)
    plt.title(f"Label: {train_labels[i]}")
    plt.show()
```

```python
drawImage(2)
```

```python
# transformation des entrées en format 4D pour le reseau convolutif
# 60000 images de 28x28 pixels avec 1 canal de couleur (gris)
train_images = train_images.reshape((60000, 28, 28, 1)).astype('float32') / 255
test_images = test_images.reshape((10000, 28, 28, 1)).astype('float32') / 255
```

```python
# construction du modèle de réseau convolutif
# 3 couches convolutives avec des couches de pooling entre elles
# une couche de flattening pour transformer les données en vecteur 1D
# 2 couches denses pour la classification
# une couche de sortie avec 10 neurones pour les 10 classes de chiffres
model = models.Sequential([
    # 1ere couche convolutive avec 32 filtres de 3x3 pixels
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    
    # 2e couche convolutive avec 64 filtres de 3x3 pixels    
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    
    # 3e couche convolutive avec 64 filtres de 3x3 pixels    
    layers.Conv2D(64, (3, 3), activation='relu'),
    
    # couche de flattening pour transformer les données en vecteur 1D
    layers.Flatten(),
    # couches denses de classification
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])
```

```python
# Compile the model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Model summary
model.summary()
```

```python
# Entrainement du modèle
history = model.fit(
    train_images, 
    train_labels, 
    epochs=5, 
    batch_size=64,
    validation_split=0.2
)
```

```python
# evaluer le modèle sur les données de test
test_loss, test_acc = model.evaluate(test_images, test_labels)
print(f'Test accuracy: {test_acc:.4f}')
```

```python
# fonction pour visualiser les prédictions
def plot_predictions(images, labels, predictions):
    plt.figure(figsize=(10, 10))
    for i in range(25):
        plt.subplot(5, 5, i+1)
        plt.xticks([])
        plt.yticks([])
        plt.grid(False)
        plt.imshow(images[i].reshape(28, 28), cmap=plt.cm.binary)
        predicted_label = np.argmax(predictions[i])
        if predicted_label == labels[i]:
            color = 'green'
        else:
            color = 'red'
        plt.title(f"P: {predicted_label}, T: {labels[i]}", color=color)
    plt.show()
```

```python
# Visualiser les prédictions sur les données de test
predictions = model.predict(test_images[25:50])
plot_predictions(test_images[25:50], test_labels[25:50], predictions)
```
